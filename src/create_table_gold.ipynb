```python
%sql
select 
  Description, 
  sum(Quantity) as sum_quantity, 
  round(avg(UnitPrice), 2) as avg_unit_price,
  round(sum(TotalPrice), 2) as total_price 
from workspace.default.online_retail_silver 
where InvoiceDate >= '2010-12-01'
group by Description
```

```python
query = """
    select 
        Description, 
        sum(Quantity) as sum_quantity, 
        round(avg(UnitPrice), 2) as avg_unit_price,
        round(sum(TotalPrice), 2) as total_price
    from workspace.default.online_retail_silver 
    where InvoiceDate >= '2010-12-01'
    group by Description
"""

df = spark.sql(query)
```
